```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Financial_Market.csv")
```

```python
df.head()
```

```python
df["Profit"] = df["Close_Price"] - df["Open_Price"]
```

```python
df["Status"] = df["Close_Price"].apply(lambda x: "Profit" if x > 0 else "Loss")
```

```python
average_close_price = df.groupby("Stock_Symbol")["Close_Price"].mean()
print(average_close_price)
```

```python
average_volume = df.groupby("Stock_Symbol")["Volume"].mean()
print(average_volume)
```

```python
sector = df.groupby("Sector")["Market_Cap"].sum()
print(sector)
```

```python
average_close_price.plot(kind="bar")
plt.title("Stock Wise Average close Price")
plt.xlabel("Stock")
plt.ylabel("Average Close Price")
```

```python
average_volume.plot(kind="bar")
plt.title("Stock Wise Average Volume")
plt.xlabel("Stock")
plt.ylabel("Volume")
plt.show()
```

```python
sector.plot(kind="pie", autopct="%1.1f%%")
plt.title("Sector Wise Market Cap Distribution")
plt.show()
```
